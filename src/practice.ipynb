```python
import pandas as pd
import requests
import json

url ="https://api.github.com/repos/microsoft/vscode/pulls"

response = requests.get(url, params={"page":1})
status= response.status_code
content_type = response.headers.get("Content-Type")
print("Status:", status)
print("Content-Type:", content_type)
print("URL:", response.url)

if status ==200:
    if "application/json" in content_type:
        data = response.json()
    elif  "text/html" in content_type:
        data = response.text()
    else:
        data= response.content
else:
    print("request failed:", status)            
print(data[:5])
```

```python
row = []

for pr in data:
    row.append({
        "pr_id": pr["id"],
    "pr_number": pr["number"],

    "author_id": pr["user"]["id"],
    "author_login": pr["user"]["login"],
    "author_association": pr["author_association"],

    "title": pr["title"],
    "state": pr["state"],
    "draft": pr["draft"],

    "created_at": pr["created_at"],
    "updated_at": pr["updated_at"],
    "closed_at": pr["closed_at"],
    "merged_at": pr["merged_at"],
    })

df = pd.DataFrame(row)
print(df)
```

```python
import requests
import  pandas as pd

url = "https://api.github.com/repos/microsoft/vscode/issues/100/timeline"

params = {
    "per_page": 100
}

headers = {
    "Accept": "application/vnd.github+json"
}

response = requests.get(
    url,
    params=params,
    headers=headers,
    timeout=30
)

print(response.status_code)

timeline = response.json()
df = pd.DataFrame(timeline)
print(df)
```

```python
for i, event in enumerate(timeline):
    print(i, event.get("event"))
```

```python
for event in timeline:
    if event.get("event") == "cross-referenced":
        print(event)
```

```python
for event in timeline:
    if "source" in event:
        print(
            "event:", event.get("event"),
            "source type:", event["source"].get("type")
        )
```

```python
for event in timeline:
    if event.get("event") == "cross-referenced":
        source_issue = event.get("source", {}).get("issue", {})

        print("Number:", source_issue.get("number"))
        print("Title:", source_issue.get("title"))
        print("Has pull_request:", "pull_request" in source_issue)
        print("Pull request data:", source_issue.get("pull_request"))
        print("---")
```
